Una librería de Python que proporciona
 estructuras de datos como Series y DataFrame,
 especialmente diseñadas para trabajar con datos
 tabulares y heterogéneos. Además, proporciona
 herramientas intuitivas para el análisis y la
 manipulación de datos.

 - Series: Es un objeto unidimensional de
 estilo array, que contiene una
 secuencia de valores del mismo tipo
 y un array asociado de etiquetas de
 datos, correspondiente a su índice.

 - DataFrame: Se corresponde con una tabla de 2
 dimensiones. Tiene un índice de fila y
 otro de columna; se podría considerar
 como un diccionario de objetos Series
 que comparten todos el mismo índice.

```python
import pandas as pd

# con un array
s = pd.Series([1, 3, 5, 6, 8])
print(s)
```

```python
# con array de numpy
import numpy as np

a = np.arange(2,11,2)
s = pd.Series(a)
print(s)
```

```python
# con un diccionario de python
altura = {'Samuel': 175,
          'Lucia': 170,
          'Julia': 172,
          'Martin': 180}
s = pd.Series(altura)
print(s)
```

```python
# con algunos elementos de un diccionario
altura = {'Samuel': 175,
          'Lucia': 170,
          'Julia': 172,
          'Martin': 180}
s = pd.Series(altura, index=['Samuel', 'Lucia'])
print(s)
```

```python
# con un escalador

s = pd.Series(5, index=['a', 'b', 'c', 'd', 'e'])
print(s)
```
