# **Creación de arrays**

```python
import numpy as np
import matplotlib.pyplot as plt
```

```python
data1 = [1,2,3]
arr_1d = np.array(data1)
print(arr_1d)
```

```python
data2 = [[1,2],[3,4]]
arr_2d = np.array(data2)
print(arr_2d)
```

```python
data3 = [[1,2,3],[4,5,6],[7,8,9]],[[1,2,3],[4,5,6],[7,8,9]],[[1,2,3],[4,5,6],[7,8,9]]
arr_3d = np.array(data3)
print(arr_3d)
```

Arrays cuyos valores son todos 0

```python
a = np.zeros((2,3))
a
```

Arrays cuyos valores son todos 1

```python
b = np.ones((2,3,4))
b
```

Arrays cuyos valores son todos, el valor indicado en el segundo parametro

```python
c = np.full((2,3),7)
c
```

Creación de un array basado en función de rangos


Ejemplo: 1= minimo, 10=maximo y 4=numero de elementos

```python
a = np.linspace(1,10,4)
a
```

```python
a2 = np.arange(9)
print(a2)
```

Ejemplo: 1=inicio, 11=para y 2=saltar los pares

```python
b= np.arange(1,11,2)
b
```

Inicialización del array con valores aleatorios

```python
d = np.random.rand(2,3,2)
print(d)
```

 Inicialización del array con valores aleatorios con una distribución normal


```python
c = np.random.randn(2,3)
print(c)
```

```python
m = np.random.randn(100000)

plt.hist(m,bins=100)
plt.show()
```
