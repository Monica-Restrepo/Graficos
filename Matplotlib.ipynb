Propiedades de plt.scatter: s = tamaño de los puntos, c = color de los puntos, marker = forma del marcador, alpha = transparencia de 0 a 1, edgecolors = color del borde del punto, linewidths = ancho del borde del punto, label = etiqueta para la leyenda

Barras verticales apiladas

bottom = especifica dónde
 empieza cada barra en el eje
 Y o en el eje X (para gráficos
 de barras horizontales). Ya
 que por defecto, las barras
 comienzan desde cero.

```python
trimestre_año = ['trimestre 1', 'trimestre 2', 'trimestre 3', 'trimestre 4']
ventas_producto1 = [300,400,500,600]
ventas_producto2 = [350,400,550,650]

plt.bar(trimestre_año, ventas_producto1,  bottom=ventas_producto2, label='Producto1', color= 'lightblue')
plt.bar(trimestre_año, ventas_producto2, label='Producto2', color= 'lightgreen')

plt.title('Ventas por trimestre peoducto 1 y 2')
plt.xlabel('Productos')
plt.ylabel('Ventas')
plt.legend()
plt.show()
```

Propiedades de plt.bar: color = color de las barras, edgecolor= color del borde de las barras, linewidth = ancho del borde de las barras, width = ancho de las barras, label = nombre para usar la leyenda, bottom = posición base de las barras

plt.tight_layout(): Evita que las etiquetas de los ejes se corten.

Ajusta los espacios entre el título, los ejes, la leyenda y las barras.

Hace que el gráfico se vea más limpio y mejor distribuido dentro del área de dibujo.

```python
categorias = ['A', 'B', 'C', 'D']
ventas = [10,20,30,40]

plt.bar(categorias,
        ventas,
        color='lightgreen',
        edgecolor='black',
        linewidth=1.5,
        width=0.6,
        label='Ventas 2025')
plt.title('Ventas por categoria')
plt.xlabel('Categorias')
plt.ylabel('Ventas')
plt.legend()
plt.grid(axis='y', linestyle='--', alpha=0.7)
plt.tight_layout()
plt.show()
```

Histogramas:  Ideal para representar datos continuos.
Permite mostrar la distribución de una variable.

np.random.randint, que es otra forma de generar números aleatorios, pero esta vez con números enteros, no continuos como en np.random.normal.

```python
# Generar datos aleatorios
edades = np.random.randint(15,90, size=100)

plt.hist(edades, bins=10, color='skyblue', edgecolor='green')
plt.title('Distribución de edades clientes')
plt.xlabel('Edades')
plt.ylabel('Numero de clientes')
plt.show()
```

Propiedades de plt.hist: bins = numero de divisiones o contenedores del histograma, color = color de las barras, edgecolor = color borde de las barras, alpha = transparencia de las barras de 0 a 1, label = etiqueta para incluir la leyenda, histtype = tipo de histograma (bar, step, stepfilled), orientation = vertical, horizontal, rwidth =  ancho relativo de las barras de 0 a 1, 1 por defecto.

np.random.normal =  estás generando números aleatorios que siguen una distribución normal (también llamada distribución gaussiana o campana de Gauss)

loc: es la media (el valor central de la distribución).

scale: es la desviación estándar (qué tan dispersos están los datos).

size: es cuántos valores quieres generar.

Es útil para:

Simular datos cuando no tienes los reales.

Crear distribuciones de prueba.

Evaluar comportamientos típicos de un fenómeno (como ventas promedio, temperaturas, etc.)

```python
data = np.random.normal(loc=50, scale=10, size=100)
plt.hist(data,
         bins=20,
         color='lightcoral',
         edgecolor='red',
         alpha=0.6,
         rwidth=0.9,
         label='Distribución de edades',
         histtype='bar')
plt.title('Distribución de edades clientes')
plt.xlabel('Edades')
plt.ylabel('Numero de clientes')
plt.legend()
plt.grid()
plt.tight_layout()
plt.show()
```

```python
data = np.random.normal(loc=50, scale=10, size=100)
plt.hist(data,
         bins=20,
         color='green',
         edgecolor='green',
         alpha=0.6,
         rwidth=0.9,
         label='Distribución de edades',
         histtype='step')
plt.title('Distribución de edades clientes')
plt.xlabel('Edades')
plt.ylabel('Numero de clientes')
plt.legend()
plt.grid()
plt.tight_layout()
plt.show()
```

```python
data = np.random.normal(loc=50, scale=10, size=100)
plt.hist(data,
         bins=20,
         color='lightblue',
         edgecolor='blue',
         alpha=0.6,
         rwidth=0.9,
         label='Distribución de edades',
         histtype='stepfilled')
plt.title('Distribución de edades clientes')
plt.xlabel('Edades')
plt.ylabel('Numero de clientes')
plt.legend()
plt.grid()
plt.tight_layout()
plt.show()
```

Gráficos de pastel: Ideal para representar proporciones de un todo.
Usualmente útil para datos categóricos

propiedades: labels = lista de nombre para cada porción, colors = lista de colores personalizados, autopct = muestra los valores porcentuales o personalizados de cada porción, startangle = ángulo inicial del gráfico en grados (gira el pastel), explode = separa porciones para destacarlas, shadow = agrega una sombra al grafico, radius = bradio del pastel, por defecto es 1, counterclock = dirección de las porciones, por defecto antihorario, pctdistance =  distancia del texto del porcentaje dede el centro, labeldistance = distancia del texto de etiquetas desde el centro.

```python
productos = ['prod1', 'prod2', 'prod3', 'prod4']
ventas = [300,400,500,600]

plt.pie(ventas, labels=productos, autopct='%1.1f%%', startangle=90)
plt.title('Proporción de ventas por producto')
plt.show()
```

```python
valores = [30, 45, 25]
etiquetas = ['ventas','Marketing','producción']
colores = ['gold','lightcoral','lightskyblue']
explode = (0.1, 0.1, 0)

plt.pie(valores,
        labels=etiquetas,
        colors=colores,
        explode=explode,
        autopct='%1.1f%%',
        shadow=True,
        startangle=80,
        counterclock=False)
plt.title('Distribución de presupuesto anual')
plt.tight_layout()
plt.show()
```

```python
import matplotlib.pyplot as plt
import numpy as np

# Crear figura y subgráficos (1 fila, 3 columnas)
fig, (ax1, ax2, ax3) = plt.subplots(1, 3, figsize=(15, 4))

# crea histograma en ax1
ax1.hist(np.random.standard_normal(100), bins=20, color='lightblue', alpha=0.6)
ax1.set_title("Histograma")

# crea gráfico de dispersión en ax2
ax2.scatter(np.arange(30), np.arange(30) + 3 * np.random.standard_normal(30))
ax2.set_title("Dispersión")

# crea gráfico en línea en ax3
ax3.plot(np.random.standard_normal(50).cumsum(), color='green')
ax3.set_title("Gráfico en línea")

# Ajusta los espacios automáticamente
plt.tight_layout()
plt.show()
```

np.random.standard_normal(50)
Genera un array de 50 números aleatorios que siguen una distribución normal estándar, es decir:

Media = 0

Desviación estándar = 1

.cumsum()
Esto significa "cumulative sum" o "suma acumulada". Toma el array anterior y suma progresivamente los valores.

útil para:

Representar fluctuaciones en precios, ventas, temperatura, etc.

Visualizar tendencias aleatorias.

```python
import matplotlib.pyplot as plt
import numpy as np

# Crear figura y subgráficos (3 filas, 1 columna)
fig, (ax1, ax2, ax3) = plt.subplots(3, 1, figsize=(6, 12))

# Histograma
ax1.hist(np.random.standard_normal(100), bins=20, color='lightblue', alpha=0.6)
ax1.set_title("Histograma")

# Dispersión
ax2.scatter(np.arange(30), np.arange(30) + 3 * np.random.standard_normal(30))
ax2.set_title("Dispersión")

# Línea
ax3.plot(np.random.standard_normal(50).cumsum(), color='green')
ax3.set_title("Gráfico en línea")

# Ajuste del espacio
plt.tight_layout()
plt.show()
```

Opciones de matplotlib.pyplot.subplots.

nrows = numero de filas de subgraficos
ncols = numero de columnas
sharex = al ajustarlo afecta todo el grafico
sharey = al ajustarlo afecta todo el grafico
wspace y hspace = controlan el porcentaje de la anchura y la altura

```python
fig, axes = plt.subplots(2, 2, sharex=True, sharey=True)
for i in range(2):
  for j in range(2):
    axes[i, j].hist(np.random.standard_normal(100), bins=20, color='lightgreen', alpha=0.6)
plt.show()
```

```python
fig, axes = plt.subplots(2, 2, sharex=True, sharey=True)
for i in range(2):
  for j in range(2):
    axes[i, j].hist(np.random.standard_normal(100), bins=20, color='lightgreen', alpha=0.6)
fig.subplots_adjust(wspace=0, hspace=0)
plt.show()
```

Anotaciones y dibujos en un subgráfico

 Las anotaciones son ideales para explicar momentos únicos.
 Los textos libres son útiles para añadir contexto.

```python
meses = np.arange(1,13)
rendimiento = [60,60,70,56,78,90,30,100,34,67,134,89]
fig, ax = plt.subplots(figsize=(10,6))
ax.plot(meses,rendimiento, marker='o', color='green', label='rendimiento')

# Anotaciones
ax.annotate('Pico maximo', xy=(10,120), xytext=(7,130), arrowprops=dict(facecolor='black', shrink=0.05), fontsize=10, horizontalalignment='left')

# Textos libres
ax.text(8,40, 'temporada baja', color='red', fontsize=8, horizontalalignment='center')

# Personalización de ejes y titulos
ax.set_xticks(meses)
ax.set_xticklabels(['Ene', 'Feb', 'Mar', 'Abr', 'May', 'Jun', 'Jul', 'Ago', 'Sep', 'Oct', 'Nov', 'Dic'])
ax.set_xlabel('Meses')
ax.set_ylabel('Rendimiento')
ax.set_title('Rendimiento Anual')
ax.legend()
plt.show()
```

Anotaciones y dibujos en un subgráfico

Con Matplolib es posible dibujar  figuras geométricas, las cuales permiten resaltar comportamientos o eventos importantes en los datos.

**Rectangulo**

```python
meses = np.arange(1,13)
produccion = [500,480,460,470,450,300,310,400,450,470,490,500]
fig, ax = plt.subplots(figsize=(10,6))
ax.plot(meses,produccion, marker='o', color='brown', label='produccion')

# Rectangulo para indicar una zona critica
rect = plt.Rectangle((5.2,250), 2,250, color='green', alpha=0.3)
ax.add_patch(rect)

# Anotaciones
ax.annotate('Problema de producción', xy=(6,300), xytext=(8,400), arrowprops=dict(facecolor='black', arrowstyle='->'), fontsize=10)

# Personalización
ax.set_xticks(meses)
ax.set_xticklabels(['Ene', 'Feb', 'Mar', 'Abr', 'May', 'Jun', 'Jul', 'Ago', 'Sep', 'Oct', 'Nov', 'Dic'])
ax.set_xlabel('Meses')
ax.set_ylabel('Producción')
ax.set_title('producción Anual')
ax.legend()
plt.show()
```

**Pentagono**

```python
meses = np.arange(1,13)
usuarios = [100, 120, 150, 180, 210, 250, 320, 400, 450, 470, 490, 500]
fig, ax = plt.subplots(figsize=(10,6))
ax.plot(meses,usuarios, marker='v', color='darkblue', label='Clientes')

# Poligono para un área en crecimiento
pgon = plt.Polygon([[5,210], [6,250], [7,320], [8,400]], color='skyblue', alpha=0.4)
ax.add_patch(pgon)

# Anotación simple
ax.text(5.5,300, 'Crecimiento acelerado', fontsize=10, color='navy')

# Personalización
ax.set_xticks(meses)
ax.set_xticklabels(['Ene', 'Feb', 'Mar', 'Abr', 'May', 'Jun', 'Jul', 'Ago', 'Sep', 'Oct', 'Nov', 'Dic'])
ax.set_xlabel('Meses')
ax.set_ylabel('Cantidad de clientes')
ax.set_title('Nuevos clientes')
ax.legend()
plt.show()
```
